```python
import outlines
from pydantic import BaseModel

class DrugInfo(BaseModel):
    patientid: str
    drug: str
    bloodsugar: str
    diabetes: str

model = outlines.models.transformer("gpt2")
generator = outlines.generate.json(model, DrugInfo)

prompt = """
Below diagnostic and drug is suggested by oncologist for a breast cancer patient based on clinical evidence
you suggest which parameters are not falling under standard threshold and need attention:

"Patient Kavita has a breast cancer and i have given her doxirubicine drug, her bloodsugar is 120 and she is diabetic"
"""

result = generator(prompt)

print(type(result))
print(result)
print(result.model_dump_json(indent=2))
```
