```python
import ollama

system_prompt = "Concise the below paragraph into 1 line."

user_prompt = """
Company XYZ shareholders and creditors approved the proposed amalgamation of RRRR
View CCC Ltd. with XYZ in June 2026. Equity shareholders reportedly
voted 99.72% in favour, while creditors voted 100% in favour.
"""

print("Sending request to Ollama...")

response = ollama.chat(
    model="llama3.2:1b",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
)

result = response["message"]["content"]

print("Result:")
print(result)
```
